```python
#Q6. Use the IQR method to detect and remove outliers from the km_driven
#column. Show before-and-after summary statistics.
#(Include your Python code
import pandas as pd

# 1. Show Summary Statistics Before Outlier Removal
print("--- SUMMARY STATISTICS BEFORE IQR CLEANING ---")
print(df['km_driven'].describe())
print(f"Original Dataset Shape: {df.shape}")

# 2. Compute IQR Boundaries for 'km_driven'
Q1 = df['km_driven'].quantile(0.25)
Q3 = df['km_driven'].quantile(0.75)
IQR = Q3 - Q1

# Define lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"\nIQR Bounds Calculated:")
print(f"Lower Bound: {lower_bound} (Values below this are outliers)")
print(f"Upper Bound: {upper_bound} (Values above this are outliers)")

# 3. Filter out the Outliers
# Keeping only data within the upper and lower bounds
df_cleaned = df[(df['km_driven'] >= lower_bound) & (df['km_driven'] <= upper_bound)]

# 4. Show Summary Statistics After Outlier Removal
print("\n--- SUMMARY STATISTICS AFTER IQR CLEANING ---")
print(df_cleaned['km_driven'].describe())
print(f"Cleaned Dataset Shape: {df_cleaned.shape}")

# 5. Calculate rows removed
rows_removed = df.shape[0] - df_cleaned.shape[0]
print(f"\nTotal outlier rows removed: {rows_removed}")
```

```python
#Q9. Generate a heatmap of the correlation matrix for all numeric columns.
#What correlations stand out the most?
#(Include your Python code
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Select only numeric columns from the DataFrame
numeric_df = df.select_dtypes(include=['number'])

# 2. Compute the correlation matrix
correlation_matrix = numeric_df.corr()

# 3. Generate the heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(
    correlation_matrix,
    annot=True,          # Display the actual correlation numbers inside the cells
    cmap='coolwarm',     # Color gradient (blue for negative, red for positive)
    fmt=".2f",          # Round numbers to two decimal places
    linewidths=0.5,      # Add subtle lines between the squares
    vmin=-1, vmax=1      # Set bounds for standard correlation scale
)

# Add titles and adjust layout
plt.title('Correlation Heatmap of Bike Features', fontsize=14, fontweight='bold')
plt.tight_layout()

# Show the plot
plt.show()
```

```python
#Q10. Summarize your findings in a brief report:
# What are the most important factors affecting a bike's selling price?
# Mention any data cleaning or feature engineering you performed.

#(Include your Python code
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual aesthetic configuration
sns.set_theme(style="whitegrid")

# =====================================================================
# STEP 1: INITIAL DATA LOADING & STRUCTURAL AUDIT
# =====================================================================
# Update with your explicit Google Drive or workspace path
file_path = '/content/drive/MyDrive/MachineLearning/BIKE DETAILS.csv'

if os.path.exists(file_path):
    df = pd.read_csv(file_path)
    print("Dataset successfully ingested.")
else:
    print("Warning: Standard path not found, initializing mock DataFrame for execution fallback.")
    # Fallback simulation mapping standard dataset specs
    import numpy as np
    np.random.seed(42)
    df = pd.DataFrame({
        'selling_price': np.random.lognormal(mean=10.8, sigma=0.5, size=500).astype(int),
        'year': np.random.randint(2010, 2024, size=500),
        'km_driven': np.append(np.random.randint(5000, 60000, size=490), np.random.randint(150000, 300000, size=10)),
        'seller_type': np.random.choice(['Individual', 'Dealer', 'Trustmark Dealer'], size=500, p=[0.7, 0.2, 0.1]),
        'owner': np.random.choice(['1st owner', '2nd owner', '3rd owner'], size=500, p=[0.7, 0.2, 0.1])
    })

print(f"Dataset Dimensions: {df.shape[0]} rows, {df.shape[1]} columns")
print("\n--- Feature Names ---")
print(df.columns.tolist())
print("\n--- Previewing First 10 Rows ---")
print(df.head(10))


# =====================================================================
# STEP 2: DATA INTEGRITY & MISSING VALUE AUDIT
# =====================================================================
print("\n=== Data Integrity Audit ===")
missing_count = df.isnull().sum()
missing_percent = (missing_count / len(df)) * 100

for col in df.columns:
    print(f"Column '{col}': {missing_count[col]} missing entries ({missing_percent[col]:.2f}%)")

# Analytical Cleaning Action (Example Execution Block)
# df['km_driven'] = df['km_driven'].fillna(df['km_driven'].median())


# =====================================================================
# STEP 3: STATISTICAL OUTLIER DETECTION & CLEANING (IQR METHOD)
# =====================================================================
print("\n=== Outlier Remediation via IQR on Mileage ===")
Q1 = df['km_driven'].quantile(0.25)
Q3 = df['km_driven'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df_cleaned = df[(df['km_driven'] >= lower_bound) & (df['km_driven'] <= upper_bound)].copy()

print(f"Lower Filter Cutoff: {lower_bound} | Upper Filter Cutoff: {upper_bound}")
print("\nSummary Statistics Before IQR Cleaning:")
print(df['km_driven'].describe())
print("\nSummary Statistics After IQR Cleaning:")
print(df_cleaned['km_driven'].describe())


# =====================================================================
# STEP 4: FEATURE ENGINEERING (ONE-HOT ENCODING CATEGORICAL DATA)
# =====================================================================
print("\n=== Executing Feature Quantization ===")
# Encodes nominal fields to integer maps
df_encoded = pd.get_dummies(df_cleaned, columns=['seller_type'], dtype=int)
print("Previewing Engineered DataFrame Fields:")
print(df_encoded.head(5))


# =====================================================================
# STEP 5: VISUALIZATIONS GENERATION LOGIC
# =====================================================================
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Plot A: Target Variable Curve Profile
sns.histplot(data=df_cleaned, x='selling_price', bins=30, kde=True, ax=axes[0, 0], color='skyblue')
axes[0, 0].set_title('Selling Price Distribution Curve', fontweight='bold')
axes[0, 0].set_xlabel('Price (INR)')

# Plot B: Categorical Valuation Delta Analysis
sns.barplot(data=df_cleaned, x='seller_type', y='selling_price', errorbar=None, ax=axes[0, 1], palette='muted')
axes[0, 1].set_title('Average Resale Value across Intermediaries', fontweight='bold')
axes[0, 1].set_ylabel('Avg Price (INR)')

# Plot C: Operational Milestones across Ownership Tiers
sns.barplot(data=df_cleaned, x='owner', y='km_driven', errorbar=None, ax=axes[1, 0], palette='viridis')
axes[1, 0].set_title('Average Usage Statistics Across Ownership Tiers', fontweight='bold')
axes[1, 0].set_ylabel('Avg Kilometers Driven')

# Plot D: Structural Variance Spread Check
sns.scatterplot(data=df_cleaned, x='year', y='selling_price', alpha=0.6, ax=axes[1, 1], color='darkblue')
axes[1, 1].set_title('Depreciation Vector: Age vs. Value', fontweight='bold')

plt.tight_layout()
plt.show()

# =====================================================================
# STEP 6: COGNITIVE INTER-CORRELATION MATRIX EXPEDITION
# =====================================================================
plt.figure(figsize=(8, 6))
numeric_matrix = df_encoded.select_dtypes(include=['number']).corr()
sns.heatmap(numeric_matrix, annot=True, cmap='coolwarm', fmt=".2f", vmin=-1, vmax=1, linewidths=0.5)
plt.title('Systemic Inter-Feature Correlation Heatmap Matrix', fontweight='bold')
plt.show()
```
